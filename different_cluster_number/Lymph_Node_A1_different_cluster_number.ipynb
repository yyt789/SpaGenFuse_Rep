```python
import pandas as pd
import matplotlib.pyplot as plt

csv_path = "../results/robustness/different_cluster_number_repeat_metrics_A1.csv"
df = pd.read_csv(csv_path)

method_order = [
    'Seurat', 'TotalVI', 'MultiVI', 'MOFA+', 'GraphST',
    'COSMOS', 'PRAGA', 'MISO', 'SpatialGlue', 'SpaGenFuse',
]
metric_cols = [
    'Acc', 'V-measure', 'Homogeneity', 'Completeness',
    'NMI', 'AMI', 'FMI', 'ARI',
]

df['Method'] = pd.Categorical(df['Method'], categories=method_order, ordered=True)
df_cluster_metrics = (
    df.groupby(['Method', 'Embedding', 'n_clusters'], observed=True, sort=False)[metric_cols]
    .mean()
    .reset_index()
    .sort_values(['Method', 'n_clusters'])
)
```

```python
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 10

fig, axes = plt.subplots(2, 4, figsize=(18, 8), sharex=True)
for ax, metric in zip(axes.flatten(), metric_cols):
    df_cluster_metrics.boxplot(
        column=metric,
        by='Method',
        ax=ax,
        grid=False,
        rot=45,
        patch_artist=True,
        boxprops={'facecolor': '#D9E8F5', 'edgecolor': '#3A5F7D'},
        medianprops={'color': '#B23A48', 'linewidth': 1.5},
        whiskerprops={'color': '#3A5F7D'},
        capprops={'color': '#3A5F7D'},
    )
    ax.set_title(metric)
    ax.set_xlabel('')
    ax.set_ylabel('Score')

fig.suptitle('')
plt.tight_layout()
plt.show()
```
