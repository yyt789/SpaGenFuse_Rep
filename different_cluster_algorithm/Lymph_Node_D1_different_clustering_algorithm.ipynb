```python
import pandas as pd
import matplotlib.pyplot as plt

csv_path = "../results/robustness/HLN_D1_different_clustering_algorithm_repeat_summary.csv"
df = pd.read_csv(csv_path)

method_order = [
    'Seurat', 'TotalVI', 'MultiVI', 'MOFA+', 'GraphST',
    'COSMOS', 'PRAGA', 'MISO', 'SpatialGlue', 'SpaGenFuse',
]
metric_cols = [
    'Acc', 'V-measure', 'Homogeneity', 'Completeness',
    'NMI', 'AMI', 'FMI', 'ARI',
]
```

```python
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 10

fig, axes = plt.subplots(2, 4, figsize=(20, 9))
for ax, metric in zip(axes.flat, metric_cols):
    values = [df.loc[df['Method'] == method, f'{metric}_mean'].to_numpy()
              for method in method_order]
    ax.boxplot(
        values,
        patch_artist=True,
        boxprops={'facecolor': '#D9E8F5', 'edgecolor': '#3A5F7D'},
        medianprops={'color': '#B23A48', 'linewidth': 1.5},
        whiskerprops={'color': '#3A5F7D'},
        capprops={'color': '#3A5F7D'},
        flierprops={'marker': 'o', 'markersize': 3,
                    'markerfacecolor': '#3A5F7D', 'markeredgecolor': '#3A5F7D'},
    )
    ax.set_xticks(range(1, len(method_order) + 1))
    ax.set_xticklabels(method_order, rotation=45)
    ax.set_title(metric)
    ax.set_ylabel('Five-repeat mean score')
fig.tight_layout()
plt.show()
```
